# MediPredict - Exploratory Data Analysis
Student Lab Notebook 01: Initial dataset checks and missing value inspection across Cleveland, PIMA, and ILPD datasets.

```python
from medipredict.data import load_clean_data, DATASET_SCHEMAS

# Load clean datasets
df_heart = load_clean_data('heart')
df_diabetes = load_clean_data('diabetes')
df_liver = load_clean_data('liver')

print('Heart shape:', df_heart.shape)
print('Diabetes shape:', df_diabetes.shape)
print('Liver shape:', df_liver.shape)
```

Found missing values in PIMA encoded as zeros for Glucose, BloodPressure, SkinThickness, Insulin, and BMI. Zero values in these lab tests are biologically impossible so we replaced them with NaNs for median imputer.

```python
# Inspect missing values count after cleaning zero placeholders
print('PIMA NaNs per column:\n', df_diabetes.isna().sum())
print('Cleveland NaNs per column:\n', df_heart.isna().sum())
print('ILPD NaNs per column:\n', df_liver.isna().sum())
```

Cleveland dataset has 303 rows with question mark '?' missing values in ca and thal columns. Class balance is reasonably equal (~54% positive).

```python
# Target balance checks
print('Heart target distribution:\n', df_heart['target'].value_counts(normalize=True))
print('Diabetes target distribution:\n', df_diabetes['Outcome'].value_counts(normalize=True))
print('Liver target distribution:\n', df_liver['Dataset'].value_counts(normalize=True))
```
