```python
import json
from pathlib import Path

# Load empirical metrics recorded from train.py
with open('../reports/metrics.json') as f:
    metrics = json.load(f)

print('Best model for Heart:', metrics['heart']['best_model_name'])
print('Best model for Diabetes:', metrics['diabetes']['best_model_name'])
print('Best model for Liver:', metrics['liver']['best_model_name'])
```

SMOTE trial observation: applying SMOTE inside CV folds did not consistently improve recall over class_weight='balanced' for Cleveland and ILPD, and slightly increased false positives. We decided to keep class_weight as primary strategy.

```python
# Compare recall with and without SMOTE for heart dataset
for name, m in metrics['heart']['models'].items():
    print(f'{name:25s} Base Recall: {m["sensitivity"]:.4f} | SMOTE Recall: {m["smote_recall"]}')
```

Boosting (AdaBoost & XGBoost) focused heavily on boundary patient samples, yielding higher sensitivity. Bagging and Random Forest gave smoother variance control.
